```python
def validate_installation(record):
    """
    Validate an installation record.
    
    Args:
        record (dict): Must contain 'Capacity_KW' and 'Status'
    
    Returns:
        bool: True if valid, False otherwise
    """
    if record.get("Capacity_KW", 0) <= 0:
        return False
    if record.get("Status") not in ["Planned", "In_Progress", "Completed"]:
        return False
    return True
```

```python
def safe_int(value, default=0):
    """Convert to int safely, return default on failure."""
    try:
        return int(value)
    except (ValueError, TypeError):
        return default


print(safe_int("123"))     # 123
print(safe_int("abc"))     # 0
print(safe_int(None))      # 0
print(safe_int("45.6"))    # 0 (not valid int) — or use float() first
```

```python
def load_capacity_from_record(record):
    try:
        return float(record["Capacity_KW"])
    except KeyError:
        print(f"Missing 'Capacity_KW' in {record.get('ID', 'unknown')}")
        return 0.0
    except ValueError:
        print(f"Invalid value in {record.get('ID')}: {record['Capacity_KW']}")
        return 0.0
```

```python
def parse_date(date_str):
    """Parse date string, return None if invalid."""
    from datetime import datetime
    try:
        return datetime.strptime(date_str, "%Y-%m-%d")
    except (ValueError, TypeError):
        return None


# Test
print(parse_date("2025-11-15"))  # datetime object
print(parse_date("2025-13-45"))  # None (invalid month/day)
print(parse_date(None))          # None
```

```python
# Cell 7 - Functions
def district_summary(installations, district_name):
    """Return count and total KW for a district."""
    filtered = [i for i in installations if i["District"] == district_name]
    return {
        "district": district_name,
        "count": len(filtered),
        "total_kw": sum(i["KW"] for i in filtered)
    }


# Use it
for d in ["Dhaka", "Sylhet", "Chattogram"]:
    summary = district_summary(installations, d)
    print(f"{summary['district']}: {summary['count']} installations, {summary['total_kw']} KW")
```

```python
# Cell 8 - try/except
def safe_float(value, default=0.0):
    """Safely convert to float."""
    try:
        return float(value)
    except (ValueError, TypeError):
        return default


print(safe_float("3.5"))       # 3.5
print(safe_float("abc"))       # 0.0
print(safe_float(None))        # 0.0
print(safe_float(5))           # 5.0
```

```python
# Cell 10 - Combine everything: district count from real data
import csv
from collections import Counter

district_counts = Counter()

with open("data/raw/customers.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        district = row["District"]
        if district:
            district_counts[district] += 1

print(f"Total unique districts: {len(district_counts)}")
print(f"\nTop 10 districts:")
for district, count in district_counts.most_common(10):
    print(f"  {district:20s} {count:>5}")
```
