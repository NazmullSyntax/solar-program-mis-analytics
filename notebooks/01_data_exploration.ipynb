```python
# List of districts
districts = ["Dhaka", "Chattogram", "Sylhet", "Rajshahi"]

# Access by index (0-based)
print(districts[0])    # "Dhaka"
print(districts[-1])   # "Rajshahi" (last)

# Length
print(len(districts))  # 4

# Add item
districts.append("Khulna")
print(districts)  # ["Dhaka", "Chattogram", "Sylhet", "Rajshahi", "Khulna"]

# Remove item
districts.remove("Sylhet")
print(districts)  # ["Dhaka", "Chattogram", "Rajshahi", "Khulna"]

# Check if exists
if "Dhaka" in districts:
    print("Found!")

# Slice
print(districts[1:3])  # ["Chattogram", "Rajshahi"]
```

```python
# Single installation record
installation = {
    "Installation_ID": "INST-00001",
    "Customer_ID": "CUST-00001",
    "District": "Dhaka",
    "Capacity_KW": 3.5,
    "Status": "Completed"
}

# Access by key (not index!)
print(installation["District"])  # "Dhaka"
print(installation["Capacity_KW"])  # 3.5

# Add new key
installation["Inspection"] = "Passed"

# Update value
installation["Status"] = "Verified"

# Check if key exists
if "Region" in installation:
    print("Has region")
else:
    print("No region key")  # এটা print হবে

# Get with default (safe)
region = installation.get("Region", "Unknown")
print(region)  # "Unknown" (crash করবে না)

# All keys, values, items
print(installation.keys())
print(installation.values())
print(installation.items())
```

```python
# 1. Square all numbers
numbers = [1, 2, 3, 4, 5]
squares = [n ** 2 for n in numbers]
print(squares)  # [1, 4, 9, 16, 25]

# 2. Filter - only even numbers
evens = [n for n in numbers if n % 2 == 0]
print(evens)  # [2, 4]

# 3. Capitalize all districts
districts = ["dhaka", "sylhet", "chattogram"]
capitalized = [d.title() for d in districts]
print(capitalized)  # ["Dhaka", "Sylhet", "Chattogram"]

# 4. From list of dicts — extract one field
installations = [
    {"ID": "INST-001", "KW": 3.5},
    {"ID": "INST-002", "KW": 5.0},
    {"ID": "INST-003", "KW": 2.0},
]
kw_values = [inst["KW"] for inst in installations]
print(kw_values)  # [3.5, 5.0, 2.0]

total_kw = sum(kw_values)
print(total_kw)  # 10.5

# 5. Filter + transform
large_installations = [
    inst["ID"] for inst in installations if inst["KW"] >= 3.0
]
print(large_installations)  # ["INST-001", "INST-002"]
```
