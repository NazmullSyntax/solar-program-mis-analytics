```python
# Count installations per district
district_counts = installations.groupby("District")["Installation_ID"].count()
print(district_counts.head())
```

```python
district_summary = installations.groupby("District").agg({
    "Installation_ID": "count",
    "Solar_Capacity_KW": ["sum", "mean", "max"],
    "Installation_Cost": "sum"
})
print(district_summary)
```

```python
district_summary = installations.groupby("District").agg(
    Total_Installs=("Installation_ID", "count"),
    Total_KW=("Solar_Capacity_KW", "sum"),
    Avg_KW=("Solar_Capacity_KW", "mean"),
    Total_Cost=("Installation_Cost", "sum")
).round(2)

print(district_summary.head())
```

```python
# Top 10 districts by installations
top_10 = district_summary.sort_values("Total_Installs", ascending=False).head(10)
print(top_10)
```

```python
# District + Status
summary = installations.groupby(["District", "Installation_Status"]).agg(
    Count=("Installation_ID", "count")
).reset_index()
print(summary.head(10))
```

```python
# Total KW
total_kw = installations["Solar_Capacity_KW"].sum()
print(f"Total: {total_kw:,.2f} KW")

# Average capacity
avg_kw = installations["Solar_Capacity_KW"].mean()
print(f"Average: {avg_kw:.2f} KW")

# Max capacity
max_kw = installations["Solar_Capacity_KW"].max()
print(f"Max: {max_kw:.2f} KW")

# Count by status
status_counts = installations["Installation_Status"].value_counts()
print(status_counts)
```

```python
def classify_capacity(kw):
    if pd.isna(kw):
        return "Unknown"
    if kw < 3:
        return "Small"
    elif kw < 10:
        return "Medium"
    else:
        return "Large"

installations["Size_Class"] = installations["Solar_Capacity_KW"].apply(classify_capacity)
print(installations[["Solar_Capacity_KW", "Size_Class"]].head(10))
```

```python
# Price per KW (BDT)
installations["Price_per_KW"] = installations.apply(
    lambda row: row["Installation_Cost"] / row["Solar_Capacity_KW"]
    if row["Solar_Capacity_KW"] > 0 else None,
    axis=1
)
```

```python
# Clean district names
customers["District_Clean"] = customers["District"].apply(
    lambda x: x.strip().title() if pd.notna(x) else None
)
```
