```python
import numpy as np

# Python list
py_list = [1, 2, 3, 4, 5]

# NumPy array
np_array = np.array([1, 2, 3, 4, 5])

print(type(np_array))  # <class 'numpy.ndarray'>
print(np_array)        # [1 2 3 4 5]
```

```python
arr = np.array([10, 20, 30, 40, 50])

# Basic stats
print(arr.mean())    # 30.0
print(arr.sum())     # 150
print(arr.min())     # 10
print(arr.max())     # 50
print(arr.std())     # 14.14...

# Generate array
print(np.arange(0, 10, 2))    # [0 2 4 6 8]
print(np.linspace(0, 1, 5))   # [0. 0.25 0.5 0.75 1.]
print(np.zeros(3))            # [0. 0. 0.]
print(np.ones(3))             # [1. 1. 1.]

# Random (reproducible)
np.random.seed(42)
print(np.random.randint(1, 100, 5))    # [52 93 15 72 61]
```

```python
import pandas as pd

# Create Series
s = pd.Series([3.5, 5.0, 2.0, 8.0], name="Capacity_KW")
print(s)
```

```python
# Method 1: From dictionary
data = {
    "Installation_ID": ["INST-001", "INST-002", "INST-003"],
    "District": ["Dhaka", "Sylhet", "Dhaka"],
    "Capacity_KW": [3.5, 5.0, 2.0],
    "Status": ["Completed", "Completed", "Pending"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Shape
print(df.shape)  # (3, 4) → 3 rows, 4 columns

# Column names
print(df.columns.tolist())  # ['Installation_ID', 'District', 'Capacity_KW', 'Status']

# Data types
print(df.dtypes)
# Installation_ID    object
# District           object
# Capacity_KW       float64
# Status             object
# dtype: object

# First few rows
print(df.head(2))

# Summary statistics
print(df.describe())

# Info
df.info()
```
