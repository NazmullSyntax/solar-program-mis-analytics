```python
import numpy as np

# Python list
py_list = [1, 2, 3, 4, 5]

# NumPy array
np_array = np.array([1, 2, 3, 4, 5])

print(type(np_array))  # <class 'numpy.ndarray'>
print(np_array)        # [1 2 3 4 5]
```

```python
result = np.array([1, 2, 3, 4, 5]) * 2
# array([ 2,  4,  6,  8, 10])
```

```python
arr = np.array([10, 20, 30, 40, 50])

# Basic stats
print(arr.mean())    # 30.0
print(arr.sum())     # 150
print(arr.min())     # 10
print(arr.max())     # 50
print(arr.std())     # 14.14...

# Generate array
print(np.arange(0, 10, 2))    # [0 2 4 6 8]
print(np.linspace(0, 1, 5))   # [0. 0.25 0.5 0.75 1.]
print(np.zeros(3))            # [0. 0. 0.]
print(np.ones(3))             # [1. 1. 1.]

# Random (reproducible)
np.random.seed(42)
print(np.random.randint(1, 100, 5))    # [52 93 15 72 61]
```
